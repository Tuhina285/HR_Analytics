```python
# Checking for outliers
numerical_col = ["Age","MonthlyIncome","TotalWorkingYears","YearsAtCompany"]

for i in numerical_col:
    plt.figure(figsize=(6,3))
    plt.boxplot(df[i])
    plt.title(i)
    plt.show()

```

```python
# Creating Age Groups
df["AgeBand"] = pd.cut(df["Age"],
                      bins= [18,25,35,45,55,60],
                      labels=["18-25","26-35","36-45","46-55","55-60"],
                      include_lowest=True
                      )
```

```python
# Creating Salary Bucket
df["SalaryBucket"] = pd.cut(df["MonthlyIncome"],
                         bins= [1000,3000,5000,8500,20000],
                         labels= ["Low","Medium","High","Very High"],
                         include_lowest= True
                           )
```

```python
# Creating Tenure Group
df["TenureGroup"] = pd.cut(df["YearsAtCompany"],
                        bins= [0,2,5,10,20,40],
                        labels= ["0-2","3-5","6-10","11-20","20-40"],
                        include_lowest= True           
                          )
```

```python
# Converting Attrition to Numeric
df["AttritionFlag"]= df["Attrition"].map({"Yes":1,"No":0})
```

```python
# Saving the file
df.to_csv("HR_Analytics_Data.csv",index= False)
```
